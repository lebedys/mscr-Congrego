```julia
# Landscape figure: four 17-row panels need more than the 160 mm \textwidth
figure_width_pt = 900

# y labels stay on every panel: dropping them widens the plot area, so the same count
# would draw a longer bar than in its neighbour
combined = (xlims=(0, xmax_mis), titlefontsize=9, tickfontsize=7,
            left_margin=2Plots.mm, right_margin=1Plots.mm,
            top_margin=8Plots.mm, bottom_margin=2Plots.mm)

# Panel (a) has its own scale because it's a smaller library
designed = merge(combined, (xlims=(0, maximum(counts_designed) * 1.16),))

# titles are bottom-aligned by leading newlines, so each carries 4 lines
fig_combined = plot(
    plot_phenotypes(labels_designed, counts_designed;
                    title="\n\n\na) Original", designed...),
    plot_phenotypes(labels_orig, counts_orig;
                    title="\n\n\nb) Assembled", combined...),
    plot_phenotypes(labels_tra, counts_tra;
                    title="\nc) Assembled\n+ pTra Disabled\n+ pTac-tSrpR Removed", combined...),
    plot_phenotypes(labels_tra_rb, counts_tra_rb;
                    title="d) Assembled\n+ pTra Disabled\n+ pTac-tSrpR Removed\n+ Roadblocking pTac",
                    combined...);
    layout=(1, 4), size=(figure_width_pt, 380), dpi=300)

savefig(fig_combined, plotsdir("ms_misassembly_phenotypes.pdf"))
savefig(fig_combined, plotsdir("ms_misassembly_phenotypes.png"))
fig_combined
```

```julia
# One row per (part set, generator): phenotype frequencies.
include(projectdir("notebooks", "phenotype_csv.jl"))

catalogues = [
    ("ms_original", pc_original),
    ("ms_roadblock_pTac", pc_roadblock),
    ("ms_disable_Tra", pc_disable_Tra),
    ("ms_disable_Tra_roadblock_pTac", pc_disable_Tra_roadblock_pTac),
]
generators = [("g", g), ("g_crypt", g_crypt)]

entries = []
for (part_set, catalogue) in catalogues, (name, generator) in generators
    entry_labels, entry_counts, _ = phenotype_distribution(generator, catalogue.inputs; pc=catalogue)
    push!(entries, (part_set, name, entry_labels, entry_counts))
end

write_phenotype_csv(datadir("ms_misassembly_phenotypes.csv"), entries)
```
