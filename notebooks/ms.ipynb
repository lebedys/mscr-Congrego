```julia
pc_filepath = scriptsdir("parts", "ms_original.jl")
include(pc_filepath)

# Reverse the spacer rename so the simulator gets real part names
rename = Dict(:BetI => :Spacer3, :LmrA => :Spacer1, :tfSp => :Spacer2, :TraRW => :TrarW)
unrename = Dict(v => k for (k, v) in rename)
unrename_full = merge(unrename, Dict(:TrarW => :TraRW))

_input_syms = pc.inputs
_labels = Symbol[pheno_labels(length(_input_syms)); :invalid]
_counts = zeros(Int, length(_labels))

for line in eachsplit(strip(assembled_circuits), "\n")
    parts = Symbol[get(unrename, Symbol(p), Symbol(p)) for p in eachsplit(line, "-")]
    circuit = Circuit(parts)
    result, _ = simulate(circuit, _input_syms; pc=pc)
    if any(ismissing, result)
        _counts[end] += 1
    else
        idx = pheno_index(Bool.(result))
        _counts[idx] += 1
    end
end

println("$(sum(_counts)) assembled circuits characterised")
using UnicodePlots: barplot
println(barplot(_labels, _counts))
```

## Roadblocking pTac

```julia
# Load roadblock pTac part catalogue (redefines module PC)
include(scriptsdir("parts", "ms_roadblock_pTac.jl"))
pc_rb = PC.pc

show(g)
summarize(g, pc_rb.inputs, pc_rb)

labels_rb, counts_rb, n_rb = phenotype_distribution(g, pc_rb.inputs; pc=pc_rb)
n_hit_rb = count(>(0), counts_rb[1:end-1])
```

```julia
plot_phenotypes(labels_rb, counts_rb)
```

```julia
# Thesis figure: \textwidth = 160 mm on A4 with 25 mm margins
textwidth_pt = 160 / 25.4 * 72

# Shared xlims: with no value axis the reader compares bar lengths across panels
xmax = maximum([counts; counts_rb]) * 1.16
panel = (xlims=(0, xmax), titlefontsize=9, tickfontsize=7,
         left_margin=2Plots.mm, right_margin=1Plots.mm,
         top_margin=2Plots.mm, bottom_margin=1Plots.mm)

fig = plot(
    plot_phenotypes(labels, counts; title="Original", panel...),
    plot_phenotypes(labels_rb, counts_rb; title="Roadblocking pTac", panel...);
    layout=(1, 2), size=(round(Int, textwidth_pt), 320), dpi=300)

savefig(fig, plotsdir("ms_phenotypes_original_vs_roadblock.pdf"))
savefig(fig, plotsdir("ms_phenotypes_original_vs_roadblock.png"))
fig
```

```julia
# size is the same generator under both catalogues: roadblocking changes what the library
# does, not how big it is. Definitions in library_stats.jl.
@assert labels == labels_rb
include(projectdir("notebooks", "library_stats.jl"))
n_ph = n_phenotypes(length(pc_rb.inputs))

print_library_stats("original", counts, n_ph)
print_library_stats("roadblocking pTac", counts_rb, n_ph)

println("\nphenotypes lost to roadblocking: "
        * join([l for (l, a, b) in zip(labels, counts, counts_rb) if a > 0 && b == 0], ", "))
```
